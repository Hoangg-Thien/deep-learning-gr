Bài 1. Tính giá trị hàm số
Yêu cầu: Hãy viết chương trình nhập vào số thực x và tính giá trị biểu thức:
f(x)=x+x55!+|x|−−√x3/2 

Moi ban nhap vao gia tri cua bien so x: 2.3
Gia tri cua ham so f(2.3) = 3.27.

```python
# Khai báo thư viện
import math

# Nhập dữ liệu
x   = float(input('Moi ban nhap vao gia tri cua bien so x: '))
# Xử lý
f_x = x + x**5 / math.factorial(5) + math.sqrt(abs(x)) / x**(3/2)

# Xuất dữ liệu
print(f'Gia tri cua ham so f({x}) = {f_x: .2f}.')
```

Bài 6. Kiểm tra số nguyên tố
Nhập vào một số nguyên dương n. Hãy kiểm tra xem n có phải là nguyên tố không?

Moi nhap nhap so nguyen n: 10
10 khong phai la so nguyen to.

Moi nhap nhap so nguyen n: 97
97 la so nguyen to.

Lưu ý: n là số nguyên tố khi và chỉ khi không tồn tại số nguyên từ  2  đến  (√n)  mà chia hết cho n.

```python
import math
# Nhập dữ liệu
n   = int(input("Moi ban nhap so nguyen n: "))

lant = True

# Xử lý 
if n <= 1:
    lant = False
else:
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            lant = False

# Xuất dữ liệu
if lant == True:
    print(f'{n} la so nguyen to.')
else:
    print(f'{n} khong la so nguyen to.')
```

Bài 7. Giải phương trình bậc 2
Nhập 3 số thực a, b, c (có kiểm tra dữ liệu nhập số thực). Hãy xây dựng hàm giải phương trình bậc 2  ax2+bx+c=0  và in kết quả giải ra màn hình.

Moi ban nhap he so a, b, c: a b c
Ban nhap sai! Moi ban nhap lai!
Moi ban nhap he so a, b, c: 1 a 3
Ban nhap sai! Moi ban nhap lai!
Moi ban nhap he so a, b, c: 1 5 6
Phuong trinh bac 2 1x^2 + 5x + 6 = 0 co 2 nghiem x1 = -2.00, x2 = -3.00.

```python
def NhapDuLieu():
    """
    Input:
        Ban phim = 1 5 6
    Output:
        (a, b, c) --> (1, 5, 6)
    """

    while True:
        try:
            a, b, c = map(float, input("Moi ban nhap he so a, b, c: ").split())
            return a, b, c

        except ValueError:
            print("Ban nhap sai! Moi ban nhap lai!")


def GiaiPhuongTrinhBac2(a, b, c):
    """
    Input: a, b, c
    Output:
        flag = -1: Vô số nghiệm
        flag = 0:  Vô nghiệm
        flag = 1:  Có 1 nghiệm
        flag = 2:  Có 2 nghiệm
    """

    import math

    flag = None
    x = ()

    # Trường hợp a = 0
    if a == 0:

        # 0x + 0 = 0
        if b == 0:

            if c == 0:
                flag = -1
                x = ()

            else:
                flag = 0
                x = ()

        # bx + c = 0
        else:
            flag = 1
            x = (-c / b,)

    # Trường hợp a khác 0
    else:

        # Tính delta
        delta = b ** 2 - 4 * a * c

        # Delta < 0
        if delta < 0:
            flag = 0
            x = ()

        # Delta = 0
        elif delta == 0:
            x1 = -b / (2 * a)

            flag = 1
            x = (x1,)

        # Delta > 0
        else:
            x1 = (-b + math.sqrt(delta)) / (2 * a)
            x2 = (-b - math.sqrt(delta)) / (2 * a)

            flag = 2
            x = (x1, x2)

    return flag, x


# ==========================
# CHƯƠNG TRÌNH CHÍNH
# ==========================

# Nhập dữ liệu
a, b, c = NhapDuLieu()

# Gọi hàm giải
flag, x = GiaiPhuongTrinhBac2(a, b, c)

# Chuỗi phương trình
s = f'Phuong trinh bac 2 {a}x^2 + {b}x + {c} = 0'

# Xuất kết quả
if flag == -1:
    print(f'{s} co vo so nghiem!')

elif flag == 0:
    print(f'{s} vo nghiem!')

elif flag == 1:
    print(f'{s} co 1 nghiem, x = {x[0]}!')

elif flag == 2:
    print(f'{s} co 2 nghiem, x1 = {x[0]}, x2 = {x[1]}!')
```

Bài 8. Các thao tác xử lý mảng số
Viết chương trình xây dựng các hàm sau cho List và kiểm tra tính đúng đắn của các hàm số:

Nhập một mảng số từ bàn phím cách nhau bởi khoảng trắng. Xuất mảng ra màn hình.
Sinh một mảng số ngẫu nhiên có n phần tử trong khoảng a, b.
Tính tổng của mảng số, đếm các phần tử chẵn, lẻ trong mảng.
Cho một mảng số, hãy trả về dãy chẳn, dãy lẻ.
a = NhapMang()
XuatMang(a)
b = SinhNgauNhien(10, -10, 10)
XuatMang(b)
tong, chan, le = TinhTongChanLe(b)
a_chan, a_le = TimChanLe(b)


```python
import random


# 1. Nhập mảng
def NhapMang():
    a = input("Nhập mảng: ")
    a = a.split()
    a = [int(x) for x in a]
    return a


# 2. Xuất mảng
def XuatMang(a):
    print("Mảng vừa nhập:", a)


# 3. Sinh mảng ngẫu nhiên
def SinhNgauNhien(n, a, b):
    mang = []

    for i in range(n):
        mang.append(random.randint(a, b))

    return mang


# 4. Tính tổng, đếm số chẵn, số lẻ
def TinhTongChanLe(a):
    tong = 0
    chan = 0
    le = 0

    for x in a:
        tong = tong + x

        if x % 2 == 0:
            chan = chan + 1
        else:
            le = le + 1

    return tong, chan, le


# 5. Tìm dãy chẵn và dãy lẻ
def TimChanLe(a):
    dãy_chan = []
    dãy_le = []

    for x in a:
        if x % 2 == 0:
            dãy_chan.append(x)
        else:
            dãy_le.append(x)

    return dãy_chan, dãy_le


# ============================
# CHƯƠNG TRÌNH CHÍNH
# ============================

# Nhập mảng từ bàn phím
a = NhapMang()

# Xuất mảng vừa nhập
XuatMang(a)


# Sinh một mảng ngẫu nhiên gồm 10 số từ -10 đến 10
b = SinhNgauNhien(10, -10, 10)

print("Mảng ngẫu nhiên:", b)


# Tính tổng, số chẵn, số lẻ của mảng b
tong, so_luong_chan, so_luong_le = TinhTongChanLe(b)

print("Tổng của mảng:", tong)
print("Số lượng số chẵn:", so_luong_chan)
print("Số lượng số lẻ:", so_luong_le)


# Tách mảng b thành 2 mảng: chẵn và lẻ
dãy_chan, dãy_le = TimChanLe(b)

print("Dãy số chẵn:", dãy_chan)
print("Dãy số lẻ:", dãy_le)
```

Bài 9. Các thao tác trên tập hợp, từ điển
Viết chương trình nhập vào một dãy ký tự a cách nhau bởi khoảng trắng. Hãy thực hiện:

In các ký tự trong a không trùng nhau và được sắp theo thứ tự tăng dần
Đếm số lần xuất hiện các ký tự trong dãy a

```python
def XuLyKhongTrung(a):
    """
    Input:
        a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
        s = ['A', 'C', 'F', 'G', 'H', 'T']
    """
    s = set([])

    for x in a:
        s.add(x)

    return sorted(s)


# XuLyKhongTrung


def DemSoLanXuatHien(a):
    """
    Input:
        a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
        dem = {'A': 2, 'G': 1, 'C': 2, 'F': 2, 'T': 1, 'H': 1}
    """
    dem = {}  # k: ky tu, v: so lan xuat hien

    for x in a:
        if x in dem:
            dem[x] += 1
        else:
            dem[x] = 1

    return dem

a = input("Nhập dãy ký tự: ").split()

print("Các ký tự không trùng nhau:", XuLyKhongTrung(a))
print("Số lần xuất hiện:", DemSoLanXuatHien(a))
```
