# Chương trình chính gọi `calc.py`

Notebook này kích hoạt file `calc.py` bằng lệnh Python. File `calc.py` nhận ba đối số:
- số thứ nhất
- phép tính `+`, `-`, `*` hoặc `/`
- số thứ hai

```python
import subprocess
import sys

calc_file = r"D:\deep-learning-gr\Tuan01\bai1\calc.py"


def run_calc(left, operation, right):
    result = subprocess.run(
        [sys.executable, calc_file, str(left), operation, str(right)],
        capture_output=True,
        text=True,
        check=True,
    )
    print(result.stdout.strip())
```

```python
# Kich hoat lenh chay calc.py
run_calc(12, "+", 5)
```

```python
# Thu them cac phep tinh khac
run_calc(20, "-", 7)
run_calc(6, "*", 4)
run_calc(15, "/", 3)
```
